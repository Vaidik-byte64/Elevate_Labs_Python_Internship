```python
#Task 1 
#Command Line Calculator

def add(x,y):
    "Add two numbers."
    return x+y

def subtract(x,y):
    "Subtract two numbers."
    return x-y

def multiply(x,y):
    "Multiplies two numbers"
    return x * y

def divide (x,y):
    "Divide two numbers, handling division by zero"
    if y == 0:
         #Raise an exception for division by zero to be caught later.
        raise ZeroDivisionError("Error: Division by zero is not allowed")
    return x/y

def main():
    print("Python Command line calculator")
    print("Enter 'quit' to exit the program.")
    print("Supported operations: +,-,*,/")

    #This loop keeps the calculator running until the user types 'quit'
    while True:
        try:
            user_input = input("Enter expression (e.g. 5 + 3): ").strip()

            if user_input.lower() == 'quit':
                print("Exiting calculator.Goodbye!")
                break

            parts = user_input.split()

            if len(parts) != 3:
                print("Invalid input format.Please enter 'number operator number'.")
                continue

            num1_str, operator, num2_str = parts

            num1 = float(num1_str)
            num2 = float(num2_str)

            result = None

            #Perform the calculation based on the operator
            if operator == "+":
                result = add(num1,num2)
            elif operator == "-":
                result = subtract(num1,num2)
            elif operator == "*":
                result = multiply(num1,num2)
            elif operator == '/':
                result = divide(num1,num2)
            else:
                print(f"Error: Unknown operator '{operator}'.")
                continue

            print(f"Result: {result}")

        except ValueError:
            print("Error: Invalid number provided.Please enter a valid number")
        except ZeroDivisionError as e:
            print(e)
        except Exception as e:
            print(f"An unexpected error occured: {e}")

if __name__ == "__main__":
    main()

             
```
